# Step 1: Transcribing the Audio

```python
# Use a pipeline as a high-level helper
from transformers import pipeline

pipe = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-medium.en",
    dtype=torch.float16,
    device='cuda',
    return_timestamps=True
)

result = pipe(audio_filename)
transcription = result["text"]
print(transcription)
open_source_transcription = transcription
```

Step 2: Processing Transcripts with LLaMA 3.2
# Prompt Engineering

```python
system_message = """
You produce minutes of meetings from transcripts, with summary, key discussion points,
takeaways and action items with owners, in markdown format without code blocks.
"""

user_prompt = f"""
Below is an extract transcript of a Denver council meeting.
Please write minutes in markdown without code blocks, including:
- a summary with attendees, location and date
- discussion points
- takeaways
- action items with owners

Transcription:
{transcription}
"""

messages = [
    {"role": "system", "content": system_message},
    {"role": "user", "content": user_prompt}
]
```

# Quantization Configuration

```python
quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)
```

# Loading & Running LLaMA 3.2

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
tokenizer = AutoTokenizer.from_pretrained(LLMA)
tokenizer.pad_token = tokenizer.eos_token
inputs_dict = tokenizer.apply_chat_template(
    messages,
    return_tensors="pt",
    padding=True,
    truncation=True
).to(device)
streamer = TextStreamer(tokenizer)
model = AutoModelForCausalLM.from_pretrained(LLMA, device_map="auto", quantization_config=quant_config)
outputs = model.generate(**inputs_dict, max_new_tokens=2000, streamer=streamer)
```

# Output Decoding & Display

```python
response = tokenizer.decode(outputs[0])
display(Markdown(response))
```
